# CNS2025 HW16 — SGD function approximation (triangle target)


```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(123)

ew = 1.0
sbs = np.arange(-10, 12, 2, dtype=float)  # -10, -8, ..., 8, 10

def basis(s, centers=sbs):
    s = np.asarray(s)
    return np.exp(-0.5 * (s[..., None] - centers[None, ...])**2)

def triangle(s):
    s = np.asarray(s)
    return 5 + s*(s < 0) - s/2

def predict(w, s):
    return basis(s) @ w
```

```python
def sgd_train(n_steps, w0=None):
    w = np.zeros(len(sbs)) if w0 is None else np.array(w0, dtype=float, copy=True)
    history = {0: w.copy()}
    for t in range(1, n_steps + 1):
        s = rng.uniform(-10, 10)
        phi = basis(s).ravel()
        yhat = float(phi @ w)
        err = float(triangle(s) - yhat)
        w = w + ew * err * phi
        if t in (20, 100):
            history[t] = w.copy()
    return history

ws = sgd_train(100)
list(ws.keys())
```

```python
ss = np.linspace(-10, 10, 401)
yt = triangle(ss)

def plot_approx(step):
    w = ws[step]
    ya = predict(w, ss)
    plt.figure(figsize=(7, 4))
    plt.plot(ss, yt, label="target")
    plt.plot(ss, ya, label=f"approx @ step {step}")
    plt.xlabel("s")
    plt.ylabel("output")
    plt.title(f"Triangle target approximation (step {step})")
    plt.legend()
    plt.show()

for step in (0, 20, 100):
    plot_approx(step)
```

```python
def plot_components(step):
    w = ws[step]
    comps = basis(ss) * w[None, :]
    plt.figure(figsize=(7, 4))
    plt.plot(ss, comps)
    plt.xlabel("s")
    plt.ylabel(r"$w_b f_b(s)$")
    plt.title(f"Components by basis function (step {step})")
    plt.show()

for step in (0, 20, 100):
    plot_components(step)
```
