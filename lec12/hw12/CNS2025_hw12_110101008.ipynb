```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (5, 3)
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
```

```python
# parameters
L = 16.0
N = 256

# spatial grid
x = np.linspace(-L, L, N, endpoint=False)

def bump(x, d):
    x = np.asarray(x)
    out = np.zeros_like(x, dtype=float)
    mask = np.abs(x) < np.abs(d)
    out[mask] = d**2 - x[mask]**2
    return out

# target function f_x(x) = bm(x,2) - bm(x,4)/8
fx = bump(x, 2.0) - bump(x, 4.0) / 8.0
```

```python
plt.figure()
plt.plot(x, fx)
plt.xlabel('x')
plt.ylabel('f_x(x)')
plt.title('Bump combination in real space')
plt.tight_layout()
plt.show()
```

```python
# Fourier transform to k-space
K = np.pi * N / (2.0 * L)
dk = 2.0 * K / N
k = np.linspace(-K, K, N, endpoint=False)

dx = 2.0 * L / N
fk = np.fft.fftshift(np.fft.fft(np.fft.fftshift(fx))) * dx

plt.figure(figsize=(6, 3))
plt.plot(k, fk.real, label='Re f_k')
plt.plot(k, fk.imag, '--', label='Im f_k')
plt.xlim(-10, 10)
plt.xlabel('k')
plt.ylabel('f_k(k)')
plt.title('Fourier transform of f_x(x)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 3))
plt.plot(k, np.abs(fk))
plt.xlim(-10, 10)
plt.xlabel('k')
plt.ylabel('|f_k(k)|')
plt.title('Magnitude of Fourier transform')
plt.tight_layout()
plt.show()
```
